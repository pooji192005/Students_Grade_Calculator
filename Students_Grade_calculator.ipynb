```python
def calculate_grade(avg):
    if avg >= 90:
        return "A"
    elif avg >=75:
        return "B"
    elif avg >= 60:
        return "C"
    elif avg >= 40 :
        return "D"
    else :
        return "fail"
subjects = int(input(f"enter marks of subject : "))
total = 0
    
for i in range(subjects):
    marks = int(input(f"enter marks of subject {i+1}:"))
    total+=marks
average = total/subjects
grade = calculate_grade(average)
print("total:",total)
print("average:",average)
print("Grade:", grade)
    
```
